# 6 模块

在 Python 中，一个 .py 文件就是一个模块。为了避免文件名冲突，Python 又引入了按目录组织模块的方法，称为包

包：普通包是包含 __init__.py 文件的文件夹，这个文件可以为空。Python 也支持不带 __init__.py 的命名空间包，初学时先使用普通包即可

层级关系：如果你有一个包叫 my_project，里面有个模块叫 utils.py，那么引用它的完整名称是 my_project.utils

## 6.1 使用模块

````python
import math #导入math模块
from math import sqrt #从math模块中导入sqrt函数
# import numpy as np #导入numpy模块并重命名为np，需要先安装第三方库

print(math.pi)
print(sqrt(9))
````

import math 后，使用 math.sqrt() 调用函数；from math import sqrt 后，可以直接使用 sqrt()

import 模块名 as 别名 可以给模块取一个简短的名称。不建议使用 from 模块名 import *，容易造成名称冲突

````python
import math as m
print(m.sqrt(16))
````

## 6.2 使用自定义模块

把常用的函数保存到一个 .py 文件中，就可以在其他文件里导入使用

下面的代码可以保存为 hello.py。直接运行这个文件时，__name__ 的值是 '__main__'；被其他文件 import 时，__name__ 的值是模块名 hello

因此，if __name__ == '__main__': 中的代码只在直接运行时执行，不会在导入时执行。Notebook 中 __name__ 通常也是 '__main__'

````python
# hello.py
def greeting(name='world'):
    return f'Hello, {name}!'

if __name__ == '__main__':
    print(greeting())
````

把另一个文件 main.py 放在 hello.py 的同一目录中，就可以这样使用：

```python
import hello

print(hello.greeting('Kan'))
```

导入模块会执行模块顶层的代码，因此把演示代码放进 __main__ 判断中。一个模块在同一个解释器进程中通常只在首次导入时执行，之后的 import 会使用缓存

## 6.3 作用域

正常的函数和变量名可以直接从模块外部引用

以一个下划线 _ 开头的函数或变量，通常表示模块内部使用，这是命名约定，并不会真正阻止外部访问

函数内部定义的局部变量只在函数内部可用，模块顶层定义的名称属于模块的全局作用域

````python
def _format_name(name):
    return name.strip().title()

def greeting(name):
    return f'Hello, {_format_name(name)}!'

print(greeting('  kan yue  '))
````

## 6.4 模块搜索路径

导入模块时，Python 会根据模块搜索机制查找模块，sys.path 列出了查找模块时使用的目录

如果出现 ModuleNotFoundError，先检查模块名、文件所在目录，以及当前使用的 Python 环境

自己的文件不要命名为 math.py、json.py 等已有模块名，避免遮蔽原来的模块

````python
import sys

print(sys.executable) # 当前使用的Python解释器
print(sys.path[:3]) # 查看部分模块搜索路径
````
