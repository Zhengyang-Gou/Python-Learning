# 7 面向对象编程

## 7.1 类和实例

类是抽象的模板，而实例是根据类创建出来的具体的对象

```python
class Student(object):
    pass
```

由于类可以起到模板的作用，因此，可以在创建实例的时候，把一些我们认为必须绑定的属性强制填写进去

```python
class Student(object):
    def __init__(self, name, score): # __init__方法的第一个参数永远是self，表示创建的实例本身，因此，在__init__方法内部，就可以把各种属性绑定到self，因为self就指向创建的实例本身
        self.name = name
        self.score = score 
```

有了__init__方法，在创建实例的时候，就不能传入空的参数了，必须传入与__init__方法匹配的参数，但self不需要传

```python
bart = Student('Bart Simpson', 59)
print(bart.name)
```

数据封装：

不应该从外部访问类内部数据，从类内部定义函数访问

```python
class Student(object):
    def __init__(self,name,score):
        self.name = name
        self.score = score

    def print_score(self):
        print(f'{self.name}: {self.score}')

    def get_grade(self):
        if self.score >= 90:
            return 'A'
        elif self.score >= 60:
            return 'B'
        else:
            return 'C'

Ky = Student('Kan Yue', 99)
Ky.print_score()
print(Ky.get_grade())
```

## 7.2 访问限制

为了保护数据，我们需要把属性隐藏起来，不让外部直接修改

实例的变量名如果以双下划线 __ 开头，就变成了一个私有变量，私有变量只有类内部可以访问，外部无法访问

```python
class Student(object):
    def __init__(self, name, grade):
        self.__name = name   # 私有属性
        self.__grade = grade # 私有属性
    
    def get_name(self):
        print(f'{self.__name}: {self.__grade}')

bart = Student('Bart Simpson', 59)
print(bart.__name) # 报错，实例不能访问私有属性
bart.get_name()    # 正常输出
```

获取和修改私有变量:

编写工具方法（Getter 和 Setter）

```python
class Student(object):
    def __init__(self, name, grade):
        self.__name = name   # 私有属性
        self.__grade = grade # 私有属性

    def get_name(self):
        return self.__name
    
    def get_grade(self):
        return self.__grade
    
    def set_grade(self, grade):
        if 0 <= grade <= 100:
            self.__grade = grade
        else:
            raise ValueError('成绩必须在0~100之间！')
```

## 7.3 继承和多态

继承：代码复用

如果这个新类和某个现有的类有很多共同点，就可以让新类继承自那个现有的类

```python
class Animals(object):
    def run(self):
        print('Animal is running...')

class Dog(Animals):
    pass # Dog类直接继承了Animals类，没有任何修改和扩展

class Cat(Animals):
    def run(self):
        print('Cat is running...') # Cat类继承了Animals类，并重写了run方法

dog = Dog()
dog.run()
cat = Cat()
cat.run()
```

多态：

同一个方法调用，作用于不同的对象，可以有不同的表现形式


```python
def run_twice(animal):
    animal.run()
    animal.run()
run_twice(Dog())
run_twice(Cat())
run_twice(Animals())
```

## 7.4 获取对象信息

使用 type

```python
# 基本类型
print(type(123))        # <class 'int'>
print(type('str'))      # <class 'str'>
print(type(None))       # <class 'NoneType'>

# 函数类型
def fn():
    pass
print(type(fn))         # <class 'function'>
```

使用 isinstance：判断继承关系

```python
class Animal(object):
    pass
class Dog(Animal):
    pass
a = Animal()
d = Dog()
print(isinstance(a, Animal)) # True
print(isinstance(d, Dog))    # True
```

使用 dir：获取所有属性和方法


```python
print(dir('abc'))
```

动态操作：hasattr、getattr、setatrr

这三个函数允许我们在不知道对象具体结构的情况下，直接操控它的内部

```python
class MyObject(object):
    def __init__(self):
        self.x = 9

obj = MyObject()
print(hasattr(obj, 'x'))  # True
print(hasattr(obj, 'y'))  # False
setattr(obj, 'y', 19)  # 设置一个属性y
print(hasattr(obj, 'y'))  # True
print(getattr(obj, 'x'))  # 9
print(getattr(obj, 'y'))  # 19
print(getattr(obj, 'z', 404))  # 获取属性z，默认返回404
```

## 7.5 实例属性和类属性

实例属性：

实例属性是绑定在具体某个实例上的，每个实例的属性都是独立的，互不影响

```python
class Student(object):
    def __init__(self, name):
        self.name = name  # 这是一个实例属性

s1 = Student('Alice')
s2 = Student('Bob')

print(s1.name)  # Alice
print(s2.name)  # Bob
```

类属性：

类属性是直接定义在类中的，它是所有实例共享的，在内存中只存在一份

```python
class Student(object):
    count = 0  # 这是一个类属性，用来统计学生人数
    
    def __init__(self, name):
        self.name = name
```

实例属性的优先级高于类属性，在编写程序的时候，千万不要把实例属性和类属性名字写成一样

```python
class Student(object):
    count = 0  # 类属性初始化

    def __init__(self, name):
        self.name = name
        Student.count += 1  # 每次创建实例，类属性自增

# 测试
a = Student('A')
b = Student('B')
c = Student('C')

print(f"目前有 {Student.count} 个学生。")  # 输出：目前有 3 个学生。
```
