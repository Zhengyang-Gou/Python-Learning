# 空间复杂度分析

[参考文章](https://programmercarl.com/algo/fundamentals/space-complexity.html)

## 概念

空间复杂度描述运行过程中占用空间随输入规模增长的趋势。分析时关注同时存在的最大占用，不是所有历史分配量之和，也不是源文件的大小

辅助空间通常不计输入数据；返回结果是否计入需要明确说明，不能在不同解法之间随意更换口径

## 示例
### 原地反转列表


```python
def reverse_in_place(nums: list[int]) -> None:
    left, right = 0, len(nums) - 1

    while left < right:
        nums[left], nums[right] = nums[right], nums[left]
        left += 1
        right -= 1
```

时间复杂度：$O(n)$

辅助空间复杂度：$O(1)$ 输入再长，也只需要固定数量的下标和临时引用，不会创建新列表

### 创建反转副本


```python
def reversed_copy(nums: list[int]) -> list[int]:
    return nums[::-1]
```

时间复杂度：$O(n)$ 切片需要复制 n 个元素引用

新增空间复杂度：$O(n)$ 返回一个新列表；如果单独排除返回结果，额外工作空间为 $O(1)$，但新列表的实际内存仍然需要存在

### 创建二维列表


```python
def make_matrix(rows: int, cols: int) -> list[list[int]]:
    return [[0] * cols for _ in range(rows)]
```

时间复杂度：$O(rows 	imes cols + rows)$ 包括建立各行，行列数都为正时通常写为 $O(rows 	imes cols)$

新增空间复杂度：$O(rows 	imes cols + rows)$ 保存全部元素引用和各行；若行列数均为 n，则为 $O(n^2)$

不要使用 `[[0] * cols] * rows` 代替：它会让多行指向同一个列表，修改一行会影响其他行

## 递归空间

递归的局部变量会随着调用一起保留，需要算入调用栈

如果每层只占常数空间，深度为 n 时栈空间为 $O(n)$；每次将规模减半时，深度通常为 $O(log n)$

如果每层还创建列表等对象，也要计算它们在同一条未返回路径上的占用，不能只数递归层数

## 易错点

- 循环执行 n 次，不代表空间一定是 $O(n)$，要看是否同时保存了 n 份数据
- `backup = nums` 只是增加引用；`backup = nums[:]` 会创建新列表
- 返回一个新数组与原地修改输入不是同一种空间开销
- 时间上的重复执行可以累加；已经释放且不再保留的数据，不能一直累加到峰值空间中
