# 时间复杂度分析

[参考文章](https://programmercarl.com/algo/fundamentals/time-complexity.html)

## 概念

时间复杂度描述主要操作次数随输入规模增长的趋势，通常记为 $O(f(n))$，不直接表示执行了多少秒

大 O 表示渐近上界，并不等于“最坏情况”本身。分析最好、平均或最坏情况时，都可以使用大 O；本章没有特别说明时讨论最坏情况，并尽量给出紧的量级

例如操作次数为 $3n^2 + 5n + 8$，规模足够大时由平方项主导，记作 $O(n^2)$

## 分析方法

- 常数项和固定系数可以省略，例如 $O(5n + 10) = O(n)$
- 顺序执行的两段代码，先将开销相加，再取主导项
- 嵌套循环要看实际执行次数，不能只按循环层数判断
- 两个独立规模要分别保留，例如遍历长度为 m 和 n 的数组是 $O(m+n)$

## 示例
### 线性遍历


```python
def total(nums: list[int]) -> int:
    result = 0

    for num in nums:
        result += num

    return result
```

设数组长度为 n，循环执行 n 次，每次只做常数次操作

时间复杂度：$O(n)$

空间复杂度：$O(1)$ 只维护一个累加变量

### 两层循环

下面统计下标满足 i < j 的组合数量，内层循环次数随 i 改变


```python
def count_pairs(n: int) -> int:
    count = 0

    for i in range(n):
        for j in range(i + 1, n):
            count += 1

    return count
```

总次数为 $(n-1)+(n-2)+cdots+1=n(n-1)/2$

时间复杂度：$O(n^2)$ 虽然不是每层都执行 n 次，但主导项仍为平方项

空间复杂度：$O(1)$

### 每次扩大两倍

下标依次取 1、2、4、8……，直到不小于 n。这里约定 n 为正整数


```python
def doubling_steps(n: int) -> int:
    index = 1
    steps = 0

    while index < n:
        index *= 2
        steps += 1

    return steps
```

执行 k 次后，index 等于 $2^k$，因此循环次数约为 $log_2 n$

时间复杂度：$O(log n)$

空间复杂度：$O(1)$

对数底数从一个固定常数换成另一个固定常数，只会产生常数系数，因此大 O 中通常省略底数

## 常见量级

| 复杂度 | 常见情形 |
| --- | --- |
| $O(1)$ | 读取列表某个下标的元素 |
| $O(log n)$ | 二分查找，每次缩小一半范围 |
| $O(n)$ | 遍历数组 |
| $O(nlog n)$ | 常见的高效比较排序 |
| $O(n^2)$ | 枚举全部下标对 |
| $O(2^n)$ | 枚举子集，尚未计入输出每个子集的成本 |
| $O(n!)$ | 枚举排列，尚未计入输出每个排列的成本 |

## 易错点

- 两个先后执行的单层循环通常是 $O(n)$，不是 $O(n^2)$
- 循环内部调用 `sum(nums)`，还需要算上这次遍历的开销
- 分支要说明输入情况，不能只看恰好提前返回的那个例子
- 复杂度相同的程序，实际运行时间也可能不同，常数和执行环境仍会影响耗时
