# 递归算法的性能分析

[参考文章](https://programmercarl.com/algo/fundamentals/recursive-time-and-space-complexity.html)

## 问题

计算斐波那契数列的第 n 项，n 为非负整数：$F(0)=0$、$F(1)=1$，之后 $F(n)=F(n-1)+F(n-2)$

示例：n = 6，返回 8，对应序列为 0、1、1、2、3、5、8

## 思路

递归写法与公式很接近，但要同时检查重复子问题和最大递归深度。以下按固定大小数值的运算模型分析

## 解法
### 直接递归


```python
class solution:
    def fib(self, n: int) -> int:
        if n <= 1:
            return n

        return self.fib(n - 1) + self.fib(n - 2)
```

例如计算 F(5) 时，F(3) 会同时出现在 F(4) 和 F(3) 两条分支中，相同子问题被反复计算

时间复杂度：指数级，可用 $O(2^n)$ 作为上界，更紧的量级为 $Theta(arphi^n)$，其中 $arphi=(1+sqrt{5})/2$

空间复杂度：$O(n)$ 两个递归分支依次执行，栈的最大深度为 n 量级，并不会同时保存整棵递归树

### 记忆化递归

用 memo 保存已经计算的结果，同一个 n 只展开一次


```python
class solution:
    def fib(self, n: int) -> int:
        memo = {0: 0, 1: 1}

        def dfs(index: int) -> int:
            if index not in memo:
                memo[index] = dfs(index - 1) + dfs(index - 2)

            return memo[index]

        return dfs(n)
```

时间复杂度：$O(n)$ 一共 n + 1 个不同状态，按字典操作的通常情况分析

空间复杂度：$O(n)$ 缓存需要线性空间，调用栈也仍然可能达到线性深度

### 迭代

只保留相邻两项，逐个向后推，不需要缓存全部状态，也不需要递归调用栈


```python
class solution:
    def fib(self, n: int) -> int:
        previous, current = 0, 1

        for _ in range(n):
            previous, current = current, previous + current

        return previous
```

时间复杂度：$O(n)$

空间复杂度：$O(1)$ 按保存固定数量数值的模型分析；如果精确计算大整数位数，保存第 n 项仍需要随 n 增长的位空间

## 另一个例子：递归二分查找

对有序且无重复元素的数组进行查找，用左右下标限定范围，避免递归时创建切片


```python
class solution:
    def search(self, nums: list[int], target: int) -> int:
        def binary_search(left: int, right: int) -> int:
            if left > right:
                return -1

            middle = left + (right - left) // 2

            if nums[middle] == target:
                return middle
            if nums[middle] > target:
                return binary_search(left, middle - 1)

            return binary_search(middle + 1, right)

        return binary_search(0, len(nums) - 1)
```

时间复杂度：$O(log n)$ 每次只进入一半区间

空间复杂度：$O(log n)$ 来自递归栈；空数组直接返回 -1

如果使用 `nums[:middle]` 等切片，复制元素的时间和同时保留切片的空间都会增加，不能直接沿用上面的结论

## 易错点

- 缓存消除了重复计算，却不一定减少最深的调用链
- 递归树节点很多，不代表它们同时占用内存
- Python 的普通函数递归不会因为写成尾递归就自动变成常数栈空间
- 长度很大的线性递归应考虑改写为循环，而不是只提高递归深度限制
