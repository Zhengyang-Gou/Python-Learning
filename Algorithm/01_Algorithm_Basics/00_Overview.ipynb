## 分析方法

- 先定义规模，例如数组长度 n、查询次数 q、两个链表长度 m 和 n
- 再数主要操作，写出随规模变化的次数
- 分别考虑循环、递归调用，以及切片、排序等隐含操作
- 最后说明时间复杂度和空间复杂度，必要时区分最坏情况、平均情况和均摊情况

**时间看总工作量，空间看同时存在的最大额外占用**

## 本章约定

没有特别说明时，复杂度按最坏情况分析，基本数值运算按常数时间估算；讨论空间时会说明是否计入返回结果

Python 整数可以任意增长，大整数的运算和存储并非恒定开销。幂、斐波那契等示例主要用于分析调用结构，暂不展开按二进制位数计算的复杂度
