# 递归算法的时间复杂度

[参考文章](https://programmercarl.com/algo/fundamentals/recursive-time-complexity.html)

## 问题

计算 $x^n$，这里 n 为非负整数，约定 n = 0 时返回 1

示例：x = 2、n = 5，返回 32；x = 3、n = 0，返回 1

## 思路

分析递归时，要看全部调用做了多少工作，不能只看有多少层。下面都按一次数值乘法为常数时间分析

## 解法
### 逐次相乘


```python
class solution:
    def power(self, x: int, n: int) -> int:
        result = 1

        for _ in range(n):
            result *= x

        return result
```

时间复杂度：$O(n)$ 做 n 次乘法

空间复杂度：$O(1)$ 按固定大小数值的分析模型估算

### 每次把指数减一

递归关系为 $x^n=xcdot x^{n-1}$，每次只递归调用一次，但需要一直减到 0


```python
class solution:
    def power(self, x: int, n: int) -> int:
        if n == 0:
            return 1

        return x * self.power(x, n - 1)
```

时间复杂度：$O(n)$ 递推式为 $T(n)=T(n-1)+O(1)$

空间复杂度：$O(n)$ 最深同时保留 n + 1 层调用

### 指数减半，但重复调用

偶数时 $x^n=(x^{n/2})^2$，奇数时再乘一个 x。下面的写法虽然每层将 n 减半，却重复计算了同一个结果


```python
class solution:
    def power(self, x: int, n: int) -> int:
        if n == 0:
            return 1

        left = self.power(x, n // 2)
        right = self.power(x, n // 2)
        result = left * right

        return result * x if n % 2 == 1 else result
```

以 n = 8 为例，每层调用数依次为 1、2、4、8、16，最后一层的 n 都为 0，合计 31 次调用

    时间复杂度：$O(n)$ 递推式为 $T(n)=2T(lfloor n/2
floor)+O(1)$，递归树的全部节点数量为线性量级

    空间复杂度：$O(log n)$ 两个子调用依次执行，峰值调用栈只是一条从根到叶子的路径

    **递归树的总节点数决定总工作量，递归深度决定这里的峰值栈空间**

    ### 只计算一次子问题：快速幂

    先把 $x^{lfloor n/2
floor}$ 保存为 half，再直接使用 half * half，消除重复递归


```python
class solution:
    def power(self, x: int, n: int) -> int:
        if n == 0:
            return 1

        half = self.power(x, n // 2)
        result = half * half

        return result * x if n % 2 == 1 else result
```

时间复杂度：$O(log n)$ 递推式变为 $T(n)=T(lfloor n/2
floor)+O(1)$，这里按 n ≥ 2 表述渐近量级

    空间复杂度：$O(log n)$ 保存递归调用栈

    ## 易错点

    - 指数减半后，要使用整数除法 `n // 2`
    - n 为奇数时不能遗漏额外的一个 x
    - 把同一个递归调用写两次，不会自动缓存结果
    - 一般不能直接用“调用次数 × 某一层的耗时”套公式，应将每次调用的局部工作相加；本例局部工作才都是常数级
    - 逐次递归的深度会随 n 增长，Python 中较大的 n 可能触及递归深度限制
