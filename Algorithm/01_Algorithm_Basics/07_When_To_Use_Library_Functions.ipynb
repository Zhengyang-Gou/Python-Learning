# 刷力扣用不用库函数？

[参考文章](https://programmercarl.com/algo/fundamentals/when-to-use-library-functions.html)

## 思路

练习一道题时，先确认它要训练的核心操作。如果正在练二分、反转、排序等算法，先手写核心逻辑，才能理解边界和不变量

如果库函数只是辅助步骤，可以使用，但仍要说明它的时间、空间开销，并遵守题目限制

## 常见操作的开销

下表按 CPython 中通常使用的容器实现讨论，n 为容器长度，k 为切片长度

| 操作 | 时间复杂度 | 需要注意 |
| --- | --- | --- |
| `len(nums)`、`nums[index]` | $O(1)$ | 针对 list 的长度和下标访问 |
| `sum(nums)`、`min(nums)`、`max(nums)` | $O(n)$ | 都需要遍历，不能因为只有一行就当作常数时间 |
| `target in nums` | $O(n)$ | 无序列表通常要逐个比较 |
| `nums[left:right]` | $O(k)$ | 创建新列表，另占 $O(k)$ 空间 |
| `nums.append(x)` | 均摊 $O(1)$ | 单次扩容可能需要复制已有引用 |
| `nums.pop(0)`、头部插入 | $O(n)$ | 后续元素引用需要移动 |
| `nums.sort()`、`sorted(nums)` | 最坏 $O(nlog n)$ | 前者修改原列表；后者创建新列表，原地排序也可能需要辅助空间 |
| dict / set 的查找 | 平均 $O(1)$，最坏可达 $O(n)$ | 不能把哈希查找描述为无条件常数时间 |

## 示例
### 手写原地反转

用双指针交换两端元素，不使用切片返回副本


```python
class solution:
    def reverseString(self, chars: list[str]) -> None:
        left, right = 0, len(chars) - 1

        while left < right:
            chars[left], chars[right] = chars[right], chars[left]
            left += 1
            right -= 1
```

时间复杂度：$O(n)$

空间复杂度：$O(1)$ 在原列表上修改，不返回新列表

`chars[::-1]` 会创建副本。`chars.reverse()` 能原地反转，但练习这道题时应先掌握双指针过程

### 用合适的容器处理队列

如果需要反复从头部取元素，`deque.popleft()` 可以避免列表 `pop(0)` 的反复搬移


```python
from collections import deque

queue = deque([1, 2, 3])
queue.append(4)
print(queue.popleft()) # 1
print(list(queue)) # [2, 3, 4]
```

deque 两端的追加和弹出为 $O(1)$；上面最后的 `list(queue)` 会遍历剩余元素，另外需要线性时间和空间

## 易错点

- 库函数缩短了代码，不代表降低了算法复杂度
- 在循环中调用 `sum()`、`index()`、`sort()` 时，要把内部工作计入总量
- Python 的 `list.sort()` 返回 None，不能写成 `nums = nums.sort()`
- 优先弄清题目的训练目标和空间要求，再决定是否调用库函数
