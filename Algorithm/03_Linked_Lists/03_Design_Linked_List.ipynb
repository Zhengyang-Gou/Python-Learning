# 设计链表
## 问题

实现 MyLinkedList，节点下标从 0 开始，不能用现成的链表类替代节点连接操作

[力扣题目链接](https://leetcode.cn/problems/design-linked-list/)

| 方法 | 作用 |
| --- | --- |
| get(index) | 返回该下标的值，下标无效时返回 -1 |
| addAtHead(val) | 在头部插入节点 |
| addAtTail(val) | 在末尾插入节点 |
| addAtIndex(index, val) | 在 index 位置之前插入，index 等于长度时追加，大于长度时不插入 |
| deleteAtIndex(index) | 删除有效下标对应的节点，下标无效时不操作 |

题目当前约束中的 index 为非负数；为兼容参考文章的扩展说明，下面也将负的插入下标按头部插入处理，get 和删除仍把负下标视为无效

## 思路

使用单链表、虚拟头结点 dummy 和长度 size。真实头结点始终为 dummy.next，dummy 不计入 size

查找第 index 个节点时，从真实头结点移动 index 次；插入或删除时，从 dummy 移动 index 次，就能找到该位置的前驱

```text
下标：             0    1    2
节点： dummy ->    1 -> 2 -> 3 -> None
```

## 解法
### 虚拟头结点 + 长度


````python
class ListNode:
    def __init__(self, val: int = 0, next=None):
        self.val = val
        self.next = next
````

````python
class MyLinkedList:
    def __init__(self):
        self.dummy = ListNode()
        self.size = 0

    def get(self, index: int) -> int:
        if index < 0 or index >= self.size:
            return -1

        current = self.dummy.next

        for _ in range(index):
            current = current.next

        return current.val

    def addAtHead(self, val: int) -> None:
        self.addAtIndex(0, val)

    def addAtTail(self, val: int) -> None:
        self.addAtIndex(self.size, val)

    def addAtIndex(self, index: int, val: int) -> None:
        if index > self.size:
            return

        index = max(0, index)
        previous = self.dummy

        for _ in range(index):
            previous = previous.next

        previous.next = ListNode(val, previous.next)
        self.size += 1

    def deleteAtIndex(self, index: int) -> None:
        if index < 0 or index >= self.size:
            return

        previous = self.dummy

        for _ in range(index):
            previous = previous.next

        previous.next = previous.next.next
        self.size -= 1
````

| 操作 | 时间复杂度 |
| --- | --- |
| get(index) | 最坏 $O(n)$ |
| addAtHead(val) | $O(1)$ |
| addAtTail(val) | $O(n)$，当前实现没有额外维护尾指针 |
| addAtIndex(index, val) | 最坏 $O(n)$，连接节点本身为 $O(1)$ |
| deleteAtIndex(index) | 最坏 $O(n)$，连接修改本身为 $O(1)$ |

空间复杂度：保存 n 个实际节点需要 $O(n)$；单次操作除新插入节点外的辅助空间为 $O(1)$

### 调用示例


````python
linked_list = MyLinkedList()
linked_list.addAtHead(1)
linked_list.addAtTail(3)
linked_list.addAtIndex(1, 2) # 1 -> 2 -> 3
print(linked_list.get(1)) # 2

linked_list.deleteAtIndex(1) # 1 -> 3
print(linked_list.get(1)) # 3
print(linked_list.size) # 2
````

### 易错点

- get、deleteAtIndex 的有效范围是 0 <= index < size
- addAtIndex 允许 index == size，不能误判为无效
- 只有真正插入或删除成功时，才修改 size
- 必须先保留原来的后继，再将前驱接到新节点上；ListNode(val, previous.next) 已经完成了这一步
- 使用 dummy 统一头部操作后，删除第 0 个节点不需要单独改写一套逻辑
