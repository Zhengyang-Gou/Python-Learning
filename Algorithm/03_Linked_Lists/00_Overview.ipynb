## 基本约定

使用 ListNode 表示单链表节点，val 保存值，next 引用下一个节点。head 为 None 表示空链表

每道题的 notebook 都提供节点定义，可以独立运行；提交力扣时，平台通常已经提供 ListNode，主要提交 solution 中对应的方法即可，平台模板使用大写 Solution 时需将类名改为 Solution

## 分析方法

- 先判断头结点是否可能变化，决定是否使用 dummy
- 明确每个指针指向当前节点、前驱节点，还是下一待处理节点
- 在覆盖 next 之前，确认后面的链表仍有引用可以到达
- 检查空链表、一个节点、连续删除和删除头结点的情况

**比较同一个节点使用 `is`；比较节点保存的数值使用 `node.val == value`**
