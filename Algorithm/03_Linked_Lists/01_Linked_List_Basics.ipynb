# 链表理论基础

[参考文章](https://programmercarl.com/algo/linked-list/linked-list-basics.html)

## 概念

链表通过节点之间的引用组织数据，节点不要求占据连续的内存位置。单链表只能沿 next 从一个节点走到下一个节点

```text
head -> [1 | next] -> [2 | next] -> [3 | None]
```

head 保存第一个节点的引用；最后一个节点的 next 为 None。head 为 None 时没有任何节点

常见类型：单链表只有 next；双链表还有 prev；循环链表的末尾连接回已有节点，不能再按“遇到 None 就结束”的方式遍历

## 节点定义


````python
class ListNode:
    def __init__(self, val: int = 0, next=None):
        self.val = val
        self.next = next
````

## 创建与遍历

下面两个辅助函数仅处理无环单链表。使用 dummy 可以统一空链表和第一个节点的连接过程


````python
def build_linked_list(values: list[int]) -> ListNode | None:
    dummy = ListNode()
    tail = dummy

    for value in values:
        tail.next = ListNode(value)
        tail = tail.next

    return dummy.next


def to_list(head: ListNode | None) -> list[int]:
    result = []
    current = head

    while current is not None:
        result.append(current.val)
        current = current.next

    return result


head = build_linked_list([1, 2, 3])
print(to_list(head)) # [1, 2, 3]
````

时间复杂度：创建和遍历分别为 $O(n)$

空间复杂度：创建链表需要 $O(n)$ 节点，to_list 需要 $O(n)$ 保存返回结果；仅沿 next 遍历本身的辅助空间为 $O(1)$

## 插入与删除

已知前驱节点 previous，在它后面插入节点，需要先让新节点连接原来的后继，再更新 previous.next

删除 previous 后面的节点，可以让 previous.next 跳过该节点。**找到前驱可能需要遍历，修改引用本身才是常数时间**


````python
head = build_linked_list([1, 3])
previous = head

new_node = ListNode(2, previous.next)
previous.next = new_node
print(to_list(head)) # [1, 2, 3]

previous.next = previous.next.next
print(to_list(head)) # [1, 3]
````

## 与数组比较

| 操作 | 动态数组，例如 Python list | 单链表 |
| --- | --- | --- |
| 按下标读取 | $O(1)$ | $O(n)$，需要从头移动 |
| 按值查找 | $O(n)$ | $O(n)$ |
| 在已知位置附近插入、删除 | 最坏 $O(n)$，需要移动引用 | 已知前驱时 $O(1)$ |
| 先按下标定位再插入、删除 | 最坏 $O(n)$ | 最坏 $O(n)$，包括定位前驱 |

Python list 内部连续保存元素引用，并不是所有元素对象本身都连续存储；链表节点还需要保存连接信息

## 易错点

- `current = current.next` 只移动当前变量，不会改变原链表的连接
- `current.next = ...` 才是在修改节点之间的结构
- 单链表删除节点时通常需要其前驱，不能把按值查找加删除一概写为 $O(1)$
- 普通无环遍历不适用于循环链表，否则可能一直无法结束
