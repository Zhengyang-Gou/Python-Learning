# 环形链表 II
## 问题

给定单链表的头结点 head，返回进入环的第一个节点；无环时返回 None，不允许修改链表

[力扣题目链接](https://leetcode.cn/problems/linked-list-cycle-ii/)

示例：head = [3,2,0,-4]，尾节点的 next 指回下标 1 的节点 2，返回这个节点对象

题面中的 pos 仅描述尾节点连接位置，不会作为参数传给方法。pos = -1 表示没有环

## 思路

可以记录访问过的节点，第一次重复访问的节点就是入口；如果要求常数额外空间，可以先用快慢指针判断有环，再利用相遇位置寻找入口

## 解法


```python
class ListNode:
    def __init__(self, val: int = 0, next=None):
        self.val = val
        self.next = next
```

### 哈希集合

从 head 沿 next 逐个访问，加入 visited。遇到一个已经在集合中的节点时立即返回；如果走到 None，则说明无环

集合保存节点对象，而不是 val，因为不同节点的值可以相同


```python
class solution:
    def detectCycle(self, head: ListNode | None) -> ListNode | None:
        visited = set()
        current = head

        while current is not None:
            if current in visited:
                return current

            visited.add(current)
            current = current.next

        return None
```

设从 head 能到达的不同节点数为 n

时间复杂度：通常为 $O(n)$，按集合查找、插入的平均常数开销分析

空间复杂度：$O(n)$ 集合最多保存全部不同节点

### 快慢指针

第一阶段：slow 每次走一步，fast 每次走两步。移动后如果 slow is fast，说明在环内相遇；如果 fast 或 fast.next 为 None，则无环

两个指针进入环后，fast 相对 slow 每次接近一个位置，因此一定会相遇

第二阶段：让一个新指针 entry 从 head 出发，另一个指针留在相遇位置。两者每次各走一步，首次相遇处就是入口

#### 为什么第二次相遇是入口？

设头结点到入口距离为 a，入口沿 next 到第一次相遇位置距离为 b，环长为 L。相遇时 slow 总共走了 t 步：

$$t=a+b+kL$$

fast 走了 2t 步，二者在环内相遇，因此它们多走的路程 t 是完整环长的倍数：$t=qL$

两式相减可得：

$$a+b=(q-k)L$$

所以从相遇位置再走 a 步，相当于从入口走了 b+a 步，恰好回到入口；同时，从 head 出发的指针也正好走到入口


```python
class solution:
    def detectCycle(self, head: ListNode | None) -> ListNode | None:
        slow = fast = head

        while fast is not None and fast.next is not None:
            slow = slow.next
            fast = fast.next.next

            if slow is fast:
                entry = head

                while entry is not slow:
                    entry = entry.next
                    slow = slow.next

                return entry

        return None
```

时间复杂度：$O(n)$ 判断环和定位入口都只需线性步数；环内相遇时间至多再增加一个环长的量级

空间复杂度：$O(1)$ 只保存固定数量的节点引用

### 易错点

- 必须先检查 fast 和 fast.next，再访问 fast.next.next
- 初始时 slow 和 fast 都是 head，应该先移动再判断相遇，不能一开始就认定有环
- 第一次相遇点通常不是入口，还要执行第二阶段
- 空链表、单节点无环、单节点自环都需要正确处理
- 不要用普通 to_list 函数打印有环链表，否则会无限遍历
