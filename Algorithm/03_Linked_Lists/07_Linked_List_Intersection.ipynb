# 链表相交
## 问题

给定两个无环单链表的头结点 headA 和 headB，返回它们第一个公共节点；没有公共节点则返回 None，且不能改变原链表结构

[力扣题目链接](https://leetcode.cn/problems/intersection-of-two-linked-lists/)

这也是参考文章中的面试题 02.07。相交指两个引用指向同一个节点对象，并不是节点值相等

```text
A: 1 -> 2 --+
            +-> 7 -> 8 -> None
B: 3 -------+
```

上图中 7、8 是同一组节点，返回公共节点 7；如果两个链表只是分别创建了值相同的节点，则不相交

## 思路

两条无环单链表一旦相交，之后的后缀会完全共享。让两个指针到链尾的剩余距离相同，再一起移动，就能同时到达第一个公共节点

## 解法


````python
class ListNode:
    def __init__(self, val: int = 0, next=None):
        self.val = val
        self.next = next
````

### 长度对齐

先分别统计长度，让较长链表的指针先走长度差。此时两个指针剩余的节点数量相同

一起向后移动，直到指向同一个节点；如果没有交点，两者会同时到达 None


````python
class solution:
    def getIntersectionNode(
        self, headA: ListNode | None, headB: ListNode | None
    ) -> ListNode | None:
        length_a = length_b = 0
        current_a, current_b = headA, headB

        while current_a is not None:
            length_a += 1
            current_a = current_a.next

        while current_b is not None:
            length_b += 1
            current_b = current_b.next

        current_a, current_b = headA, headB

        if length_a < length_b:
            current_a, current_b = current_b, current_a
            length_a, length_b = length_b, length_a

        for _ in range(length_a - length_b):
            current_a = current_a.next

        while current_a is not current_b:
            current_a = current_a.next
            current_b = current_b.next

        return current_a
````

设两条链表长度为 m 和 n

时间复杂度：$O(m+n)$ 统计长度和同步遍历都只需线性时间

空间复杂度：$O(1)$ 不保存节点集合，也不改动 next

### 双指针切换链表

pointer_a 从 A 出发，走到 None 后改从 B 出发；pointer_b 从 B 出发，走到 None 后改从 A 出发

两个指针都会经过两条链表的非公共部分，相当于消除了长度差。有交点时会在公共部分的入口相遇，没有交点时最终同时为 None

设 A 独有部分长 a、B 独有部分长 b、公共部分长 c。切换后到达入口的路径都包含 a+b+c 个节点步长，空引用切换产生的额外一步也相同，因此能够对齐


````python
class solution:
    def getIntersectionNode(
        self, headA: ListNode | None, headB: ListNode | None
    ) -> ListNode | None:
        pointer_a, pointer_b = headA, headB

        while pointer_a is not pointer_b:
            pointer_a = pointer_a.next if pointer_a is not None else headB
            pointer_b = pointer_b.next if pointer_b is not None else headA

        return pointer_a
````

时间复杂度：$O(m+n)$ 每个指针最多走过两条链表和常数次切换

空间复杂度：$O(1)$

### 易错点

- 使用 `is` 比较节点身份，不能用两个 val 是否相等判断相交
- 返回的是节点对象，不是这个节点的值
- 题目保证无环；有环链表不能直接套用这两种终止逻辑
- 不能为了构造交点而临时修改 next，这会改变输入结构
