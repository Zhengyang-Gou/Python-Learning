# 翻转链表
## 问题

给定无环单链表的头结点 head，反转节点的 next 方向，返回新的头结点

[力扣题目链接](https://leetcode.cn/problems/reverse-linked-list/)

示例：head = [1,2,3,4,5]，返回 [5,4,3,2,1]；空链表仍返回空链表

## 思路

节点本身不变，修改的是连接方向。遍历时既要把当前节点接到已反转部分，又要保留通向未处理部分的引用

## 解法


```python
class ListNode:
    def __init__(self, val: int = 0, next=None):
        self.val = val
        self.next = next
```

### 双指针迭代

previous 指向已经反转部分的头结点，初始为 None；current 指向下一个待反转节点，初始为 head

每次按顺序执行：保存 current.next、将 current.next 指向 previous、移动 previous、移动 current

以 [1,2,3] 为例：

| 处理完成的节点 | previous 所在的已反转部分 | current 所在的未处理部分 |
| --- | --- | --- |
| 尚未处理 | None | 1 -> 2 -> 3 -> None |
| 1 | 1 -> None | 2 -> 3 -> None |
| 2 | 2 -> 1 -> None | 3 -> None |
| 3 | 3 -> 2 -> 1 -> None | None |

循环结束时 current 为 None，previous 就是新头结点


```python
class solution:
    def reverseList(self, head: ListNode | None) -> ListNode | None:
        previous = None
        current = head

        while current is not None:
            next_node = current.next # 先保存原来的后继
            current.next = previous
            previous = current
            current = next_node

        return previous
```

时间复杂度：$O(n)$ 每个节点只处理一次

空间复杂度：$O(1)$ 只维护固定数量的引用，不创建新链表

### 递归

先递归反转 head 后面的链表，返回的新头结点保存为 new_head。此时 head.next 仍指向原来的第二个节点，它已经成为反转后子链表的尾节点

用 head.next.next = head，将这个尾节点接回 head，再令 head.next = None，断开旧的正向连接

空链表或只有一个节点时直接返回，是递归终止条件


```python
class solution:
    def reverseList(self, head: ListNode | None) -> ListNode | None:
        if head is None or head.next is None:
            return head

        new_head = self.reverseList(head.next)
        head.next.next = head
        head.next = None

        return new_head
```

时间复杂度：$O(n)$

空间复杂度：$O(n)$ 递归深度随链表长度增长；较长链表可能触及 Python 的递归深度限制，实际使用时优先考虑迭代

### 易错点

- 覆盖 current.next 前，先保存后继，否则会丢失未处理部分
- 迭代结束返回 previous，不能返回已经为 None 的 current
- 递归回接后必须断开 head.next，否则会形成环
- 只反转节点的值与真正反转节点连接不是同一种操作
