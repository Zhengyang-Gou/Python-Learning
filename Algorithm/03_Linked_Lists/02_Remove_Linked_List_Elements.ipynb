# 移除链表元素
## 问题

给定无环单链表的头结点 head 和整数 val，移除所有值等于 val 的节点，返回处理后的头结点

[力扣题目链接](https://leetcode.cn/problems/remove-linked-list-elements/)

示例 1：head = [1,2,6,3,4,5,6]，val = 6，返回 [1,2,3,4,5]

示例 2：head = [7,7,7]，val = 7，返回空链表；空链表输入也返回空链表

## 思路

删除节点不需要移动后面的全部节点，只需要让前驱的 next 跳过它。头结点没有前驱，因此可以单独处理，也可以引入虚拟头结点统一操作

## 解法

先定义节点，后面两种实现共用这个定义


````python
class ListNode:
    def __init__(self, val: int = 0, next=None):
        self.val = val
        self.next = next
````

### 单独处理头结点

先用 while 连续移除值等于 val 的头结点。剩下的头结点一定需要保留，再从它开始检查下一个节点

如果删除了 current.next，current 保持不动，继续检查新接上的后继；否则 current 才向后移动


````python
class solution:
    def removeElements(self, head: ListNode | None, val: int) -> ListNode | None:
        while head is not None and head.val == val:
            head = head.next

        current = head

        while current is not None and current.next is not None:
            if current.next.val == val:
                current.next = current.next.next
            else:
                current = current.next

        return head
````

时间复杂度：$O(n)$ 每个节点只处理一次

空间复杂度：$O(1)$ 只移动引用，不创建结果链表

### 虚拟头结点

新建 dummy，让 dummy.next 指向 head，从 dummy 开始检查后继。这样实际头结点也有了前驱，可以使用同一套删除逻辑

```text
dummy -> 6 -> 6 -> 2 -> None
   删除后继，current 不动
dummy -> 6 -> 2 -> None
   继续删除后继
dummy -> 2 -> None
```

返回 dummy.next，它才是处理后链表的实际头结点


````python
class solution:
    def removeElements(self, head: ListNode | None, val: int) -> ListNode | None:
        dummy = ListNode(0, head)
        current = dummy

        while current.next is not None:
            if current.next.val == val:
                current.next = current.next.next
            else:
                current = current.next

        return dummy.next
````

时间复杂度：$O(n)$

空间复杂度：$O(1)$ 只增加一个虚拟节点和固定数量的引用

### 易错点

- 头部可能连续出现多个待删除节点，第一种方法不能只用一次 if
- 删除后立刻移动 current，会漏掉连续的待删除节点
- 空链表不能直接访问 head.val，先判断是否为 None
- dummy 的值不参与判断，最终返回 dummy.next
