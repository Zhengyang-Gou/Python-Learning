# 删除链表的倒数第 N 个节点
## 问题

给定非空无环链表 head，删除倒数第 n 个节点，返回处理后的头结点。题目保证 1 <= n <= 链表长度

[力扣题目链接](https://leetcode.cn/problems/remove-nth-node-from-end-of-list/)

示例 1：head = [1,2,3,4,5]，n = 2，返回 [1,2,3,5]

示例 2：head = [1]，n = 1，返回空链表

示例 3：head = [1,2]，n = 2，返回 [2]，说明实际头结点也可能被删除

## 思路

如果链表长度为 length，倒数第 n 个节点的正向下标为 length - n。删除时要找到它的前驱，因此引入 dummy

也可以让两个指针保持固定间隔，用一次扫描找到前驱，避免先单独统计长度

## 解法


```python
class ListNode:
    def __init__(self, val: int = 0, next=None):
        self.val = val
        self.next = next
```

### 两次遍历

第一遍得到 length；第二遍从 dummy 向后移动 length - n 次，停在待删除节点的前驱


```python
class solution:
    def removeNthFromEnd(self, head: ListNode, n: int) -> ListNode | None:
        length = 0
        current = head

        while current is not None:
            length += 1
            current = current.next

        dummy = ListNode(0, head)
        previous = dummy

        for _ in range(length - n):
            previous = previous.next

        previous.next = previous.next.next
        return dummy.next
```

时间复杂度：$O(length)$ 两次顺序遍历，常数系数不会改变量级

空间复杂度：$O(1)$

### 快慢指针

fast 和 slow 都从 dummy 开始，让 fast 先走 n + 1 步。之后两个指针每次都向后移动一步，直到 fast 为 None

**这里选择的是 fast 停在 None、slow 停在待删节点前驱的约定**，因此提前移动的距离为 n + 1

以 [1,2,3,4,5]、n = 2 为例：

| 阶段 | slow | fast |
| --- | --- | --- |
| 初始 | dummy | dummy |
| fast 先走 3 步 | dummy | 3 |
| 同时走一步 | 1 | 4 |
| 同时走一步 | 2 | 5 |
| 同时走一步 | 3 | None |

此时删除 slow.next，也就是节点 4


```python
class solution:
    def removeNthFromEnd(self, head: ListNode, n: int) -> ListNode | None:
        dummy = ListNode(0, head)
        slow = fast = dummy

        for _ in range(n + 1):
            fast = fast.next

        while fast is not None:
            fast = fast.next
            slow = slow.next

        slow.next = slow.next.next
        return dummy.next
```

时间复杂度：$O(length)$ fast 沿链表前进，slow 只走其中一部分，不需要先做一遍长度统计

空间复杂度：$O(1)$

### 易错点

- 返回 dummy.next，才能正确处理删除原头结点的情况
- 提前走 n 步还是 n + 1 步，要和停止条件配套，不能混用不同版本
- 这里题目保证 n 有效，因此 fast 提前移动时不会从 None 再取 next
- 只有一个节点且 n = 1 时，slow 停在 dummy，删除后 dummy.next 为 None
