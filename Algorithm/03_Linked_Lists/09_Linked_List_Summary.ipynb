# 链表总结篇

[参考文章](https://programmercarl.com/algo/linked-list/linked-list-summary.html)

## 常见题目与方法

| 题目 | 关键思路 | 时间复杂度 | 额外空间 |
| --- | --- | --- | --- |
| 移除链表元素 | dummy 统一头部删除，删除后前驱不动 | $O(n)$ | $O(1)$ |
| 设计链表 | dummy + size，先定位前驱再修改连接 | 单次最坏 $O(n)$ | 单次辅助 $O(1)$，节点总存储 $O(n)$ |
| 翻转链表 | 保存 next，再修改方向 | $O(n)$ | 迭代 $O(1)$，递归 $O(n)$ |
| 删除倒数第 n 个节点 | fast 先走 n+1 步，slow 最后停在前驱 | $O(length)$ | $O(1)$ |
| 链表相交 | 长度对齐，或走完后切换链表 | $O(m+n)$ | $O(1)$ |
| 环形链表 II | 快慢指针相遇，再从 head 和相遇点同步前进 | $O(n)$ | $O(1)$ |

设计链表中头插为常数时间；表中的最坏线性时间主要来自按下标定位。环形链表中 n 指可达的不同节点数

## 虚拟头结点

```text
dummy -> 原头结点 -> ... -> None
```

头部可能被删除或插入时，dummy 为第一个真实节点补充了前驱，减少特殊分支。dummy 不属于题目结果，最后返回 dummy.next

删除节点的核心是找到前驱 previous，再修改 previous.next；不是把当前变量重新赋值就算删除

## 三种双指针要区分

| 场景 | 两个指针的关系 | 停止条件 |
| --- | --- | --- |
| 翻转链表 | 一个指向已反转部分，一个指向未处理部分 | current 为 None |
| 删除倒数节点 | 保持固定的 next 步数间隔 | fast 为 None 时 slow 在待删节点前面 |
| 判断环 | 一个每次走一步，一个每次走两步 | 相遇，或 fast 无法继续走两步 |

名字都可以叫快慢指针，但每道题维护的关系不同，移动距离和结束条件必须一起推导

## 引用操作的检查顺序

- 修改前：当前节点是否为 None？原来的后继后面还要不要访问？
- 修改时：保存后继、建立新连接，再断开或覆盖旧连接
- 修改后：指针应该前进，还是留在原处继续检查新接上的节点？
- 返回时：真正的新头结点是 head、dummy.next，还是 previous？

**相交和相遇都比较节点身份，不能用节点数值相等代替**

## 边界情况

| 输入 | 重点检查 |
| --- | --- |
| 空链表 | 不访问 None.val 或 None.next |
| 一个节点 | 删除后是否变成 None，反转后是否保持自身 |
| 所有节点都应删除 | 是否能够连续处理头部并正确返回 None |
| 删除头结点、尾结点 | 下标、提前移动距离和最终返回值是否正确 |
| 两条链表值相同但节点不同 | 必须返回没有交点 |
| 入口就是头结点、单节点自环 | 第二阶段是否正确返回入口 |

## 复习顺序

先能独立完成节点定义、创建、遍历和设计链表，再复习原地反转、固定间隔双指针，最后推导环入口的距离关系

每道题都可以先画一个两三个节点的例子，逐行跟踪引用变化。复杂度分析要区分“寻找节点”和“修改连接”，递归版本也要算调用栈
