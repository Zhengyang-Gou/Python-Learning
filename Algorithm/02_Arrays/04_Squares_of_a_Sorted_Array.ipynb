# 有序数组的平方
## 问题
给你一个按非递减顺序排序的整数数组 nums，返回每个数字的平方组成的新数组，要求也按非递减顺序排序

[力扣题目链接](https://leetcode.cn/problems/squares-of-a-sorted-array/)

示例 1:
- 输入：nums = [-4,-1,0,3,10]
- 输出：[0,1,9,16,100]
- 解释：平方后，数组变为 [16,1,0,9,100]，排序后，数组变为 [0,1,9,16,100]

示例 2:
- 输入：nums = [-7,-3,2,3,11]
- 输出：[4,9,9,49,121]

## 思路
原数组有序，但平方后不一定有序，因为负数的绝对值可能比正数大。例如 -4 < -1，但平方后 16 > 1

最直接的方法是先将每个元素平方，再对新数组排序

如果利用原数组已经有序这一条件，就可以发现：在当前区间内，平方后的最大值一定在最左端或最右端。因此可以用两个指针比较两端元素的平方，将较大的值从后向前填入结果数组

## 解法
### 暴力求解
遍历数组，将每个元素的平方存入新数组，再对新数组进行升序排序


```python
class solution:
    def sortedSquares(self, nums: list[int]) -> list[int]:
        result = [num * num for num in nums]
        result.sort()

        return result
```

时间复杂度：$O(n \log n)$ 平方需要遍历一次数组，排序的时间复杂度为 $O(n \log n)$

空间复杂度：$O(n)$ 需要保存平方后的新数组，Python 排序最坏情况下也需要 $O(n)$ 辅助空间


### 双指针法
定义两个指针：

左指针 left：指向当前区间的最左端，初始值为 0

右指针 right：指向当前区间的最右端，初始值为 len(nums) - 1

创建长度为 n 的结果数组 result，用下标 index 指向待填入的位置，初始值为 n - 1

每次比较 nums[left] 和 nums[right] 的平方：
- 如果左端的平方更大，将它填入 result[index]，然后 left 向右移动一位
- 否则，将右端的平方填入 result[index]，然后 right 向左移动一位；两端平方相等时，选择任意一端都可以
- 填入一个元素后，index 向前移动一位

因为每次取出的都是当前区间中最大的平方值，所以要从结果数组的末尾开始填，最终得到非递减顺序的数组

循环条件使用 left <= right，因为 left == right 时还有最后一个元素需要处理

以 nums = [-4,-1,0,3,10] 为例，依次填入的值为 100、16、9、1、0，填入位置从下标 4 到下标 0，最终结果为 [0,1,9,16,100]


```python
class solution:
    def sortedSquares(self, nums: list[int]) -> list[int]:
        left, right = 0, len(nums) - 1
        result = [0] * len(nums)
        index = len(nums) - 1

        while left <= right:
            left_square = nums[left] * nums[left]
            right_square = nums[right] * nums[right]

            if left_square > right_square:
                result[index] = left_square
                left += 1

            else:
                result[index] = right_square
                right -= 1

            index -= 1

        return result
```

时间复杂度：$O(n)$ 每次循环处理一个元素，两个指针总共移动 n 次

空间复杂度：$O(n)$ 需要一个长度为 n 的结果数组；如果不计返回结果占用的空间，额外空间为 $O(1)$
